```python
import json

# -------------------------------------------------
# Funcion que recibe una lista y genera un diccionario
# -------------------------------------------------
def lista_a_diccionario(datos):
    """
    Recibe una lista con 3 elementos:
    [nombre, apellido, edad]
    y devuelve un diccionario.
    """

    if not isinstance(datos, list):
        raise TypeError("La entrada debe ser una lista.")

    if len(datos) != 3:
        raise ValueError("La lista debe contener exactamente 3 elementos: nombre, apellido y edad.")

    nombre, apellido, edad = datos

    if not isinstance(nombre, str) or not nombre.strip():
        raise ValueError("El nombre debe ser una cadena de texto valida.")

    if not isinstance(apellido, str) or not apellido.strip():
        raise ValueError("El apellido debe ser una cadena de texto valida.")

    if not isinstance(edad, int) or edad < 0:
        raise ValueError("La edad debe ser un numero entero positivo.")

    persona = {
        "nombre": nombre.strip(),
        "apellido": apellido.strip(),
        "edad": edad
    }

    return persona


# -------------------------------------------------
# Clase Persona
# -------------------------------------------------
class Persona:
    def __init__(self, nombre, apellido, edad):
        self.nombre = nombre
        self.apellido = apellido
        self.edad = edad

    def mostrar_datos(self):
        return f"Nombre: {self.nombre}, Apellido: {self.apellido}, Edad: {self.edad}"


# -------------------------------------------------
# Programa principal
# -------------------------------------------------

# Tres listas con datos ficticios de posibles personas reales
datos1 = ["Carlos", "Ramirez", 28]
datos2 = ["Ana", "Lopez", 22]
datos3 = ["Luis", "Martinez", 35]

# Ejecutar la funcion tres veces
persona1 = lista_a_diccionario(datos1)
persona2 = lista_a_diccionario(datos2)
persona3 = lista_a_diccionario(datos3)

# Almacenar resultados en la estructura mas adecuada
# En este caso, una lista de diccionarios
personas = [persona1, persona2, persona3]

print("Resultados almacenados en una lista de diccionarios:")
for p in personas:
    print(p)

# -------------------------------------------------
# Guardar en archivo JSON
# -------------------------------------------------
nombre_archivo = "personas.json"

with open(nombre_archivo, "w", encoding="utf-8") as archivo:
    json.dump(personas, archivo, indent=4, ensure_ascii=False)

print("\nLos datos se guardaron correctamente en el archivo personas.json")

# -------------------------------------------------
# Leer y mostrar el contenido del archivo JSON
# -------------------------------------------------
with open(nombre_archivo, "r", encoding="utf-8") as archivo:
    datos_leidos = json.load(archivo)

print("\nContenido leido desde el archivo JSON:")
for persona in datos_leidos:
    print(persona)

# -------------------------------------------------
# Crear una instancia de la clase Persona con datos propios
# -------------------------------------------------
mi_persona = Persona("Alejandro", "Pacheco", 30)

print("\nInstancia de la clase Persona:")
print(mi_persona.mostrar_datos())
```
